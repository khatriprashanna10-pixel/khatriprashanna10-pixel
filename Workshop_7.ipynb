Example 1: We will start with a simple problem. Consider the following linear systems:
a. 𝑥 + 3𝑦 = 5

𝑥 − 2𝑦 =− 5

𝑥 + 𝑦 = 3

b. 2𝑥 + 3𝑦 = 3

4𝑥 + 6𝑦 = 6

3𝑥 − 2𝑦 = 1

Solve the linear systems graphical


```python
import numpy as np
import matplotlib.pyplot as plt
""" Part 1 - Problem 1: Solve linear systems graphically. """
x = np.linspace(-5,5,100)
y1 = (5-x)/3
y2 = (5+x)/2
y3 = 3-x
fig, ax = plt.subplots(figsize=(6,6))
ax.plot(x,y1,'r',label='x + 3y = 5')
ax.plot(x,y2,'g',label='x - 2y = -5')
ax.plot(x,y3,'b',label='x + y = 3')
ax.set_xlim(-4,4)
ax.set_ylim(-2,6)
ax.axvline(color='black')
ax.axhline(color='black')
ax.grid(True)
ax.legend()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear System A")
plt.show()
```

The three lines intersect at one common point (1, 2). Therefore, the system has one unique solution.

Task 1:

Plot the graph for linear system b and write the point of intersection between three lines

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5, 5, 100)
y1 = (3 - 2*x) / 3
y2 = (6 - 4*x) / 6
y3 = (3*x - 1) / 2
plt.figure(figsize=(6,6))
plt.plot(x, y1, 'r', label='2x + 3y = 3')
plt.plot(x, y2, 'g', label='4x + 6y = 6')
plt.plot(x, y3, 'b', label='3x - 2y = 1')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Task 1: Graph of Linear System (b)")
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

The first and second equations represent the same line because the second equation is exactly twice the first. The third equation intersects this line at the point (1, 1/3). Therefore, the three equations have a common intersection point (1, 0.3333).

4. linalg.solve(a,b): solve a linear system of equations ax=b

```python
import numpy as np
# Coefficient matrix
A = np.array([[1, 2],
              [3, 4]])
# Constant vector
b = np.array([1, 2])
# Solve Ax = b
x = np.linalg.solve(A, b)
print("Solution:")
print(x)
```

5. linalg.pinv(a): compute the pseudo-inverse of a matrix

```python
import numpy as np
# Create a random 3x2 matrix
A = np.random.randn(3, 2)
# Compute pseudo-inverse
A_pinv = np.linalg.pinv(A)
print("Matrix A:")
print(A)
print("\nPseudo-inverse of A:")
print(A_pinv)
# Verify the pseudo-inverse property
print("\nCheck:")
print(np.allclose(A, A @ A_pinv @ A))
```

 is the pseudo-inverse of A. The np.allclose() function checks whether the two matrices are equal within a small numerical tolerance.

Task 3:
1. Create a matrix 𝑨 of order 5 × 5 with random entries. Calculate the rank of the matrix 𝑨
to check the number of linearly independent columns or rows and to check if the matrix is
invertible. What are the other ways to check if the matrix is invertible?

```python
import numpy as np
# Create a 5x5 random matrix
A = np.random.randint(1, 10, (5, 5))
print("Matrix A:")
print(A)
# Calculate the rank
rank = np.linalg.matrix_rank(A)
print("\nRank of Matrix A:", rank)
# Check if the matrix is invertible
if rank == 5:
    print("The matrix is invertible.")
else:
    print("The matrix is NOT invertible.")
# Another way to check invertibility
det = np.linalg.det(A)
print("\nDeterminant:", det)
if det != 0:
    print("The determinant is non-zero, so the matrix is invertible.")
else:
    print("The determinant is zero, so the matrix is not invertible.")
```

2. Generate a matrix 𝑨 of order 5 × 5 with random entries and generate a vector (column) 𝒙 of
length 5 with random entries. Calculate Matrix-vector product 𝒃 = 𝑨𝒙. Now, we have a linear
system, i.e. the matrix 𝑨 and a vector 𝒃. Solve this Linear System.

```python
import numpy as np
A = np.random.randint(1, 10, (5, 5))
x = np.random.randint(1, 10, 5)
b = A @ x
print("Matrix A:")
print(A)
print("\nVector x:")
print(x)
print("\nVector b = A @ x:")
print(b)
solution = np.linalg.solve(A, b)
print("\nRecovered x:")
print(solution)
```

3. Solve the following system.

```python
import numpy as np
A = np.array([
    [8, 3, -2],
    [-4, 7, 5],
    [3, 4, -12]
])
B = np.array([9, 15, 35])
solution = np.linalg.solve(A, B)
print("Solution:")
print("x =", solution[0])
print("y =", solution[1])
print("z =", solution[2])
```
