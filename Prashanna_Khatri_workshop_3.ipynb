1. Plot a circle with center at (a, b) and radius r. The values of a, b, and r can be chosen appropriately.

```python
import numpy as np
import matplotlib.pyplot as plt

# Center of the circle
a = 2
b = 3

# Radius
r = 5

# Angle from 0 to 2π
theta = np.linspace(0, 2 * np.pi, 500)

# Parametric equations
x = r * np.cos(theta) + a
y = r * np.sin(theta) + b

# Plot the circle
plt.plot(x, y, label="Circle")
plt.scatter(a, b, color="red", label="Center")

# Make the circle look circular
plt.axis('equal')

# Labels and title
plt.title("Circle with Center (a, b) and Radius r")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()

plt.show()
```

2.Plot an Ellipse ( ((x-p)^2)/a^2) +( ((y-q)^2)/b^2) =1 with center at (p, q), length of minor axis 2a and major axis 2b. The values of p, q, a, and b can be chosen appropriately.

```python
import numpy as np
import matplotlib.pyplot as plt

# Center of the ellipse
p = 2
q = 3

# Semi-minor axis and semi-major axis
a = 4
b = 6

# Angle from 0 to 2π
theta = np.linspace(0, 2 * np.pi, 500)

# Parametric equations of the ellipse
x = p + a * np.cos(theta)
y = q + b * np.sin(theta)

# Plot the ellipse
plt.plot(x, y, label="Ellipse")
plt.scatter(p, q, color="red", label="Center")

# Make the scaling equal
plt.axis("equal")

# Labels and title
plt.title("Ellipse with Center (p, q)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()

plt.show()
```

3. Plot a Parabola (𝑦 − 𝑝) ^2= 4𝑎(𝑥 − 𝑞) with vertex at (𝑝,𝑞). The values of p, q, and a can be chosen appropriately.

```python
import numpy as np
import matplotlib.pyplot as plt

# Vertex of the parabola
p = 2
q = 1

# Distance from vertex to focus
a = 2

# Generate y-values
y = np.linspace(-10, 10, 500)

# Calculate x-values
x = ((y - p) ** 2) / (4 * a) + q

# Plot the parabola
plt.plot(x, y, label="Parabola")
plt.scatter(q, p, color="red", label="Vertex")

# Labels and title
plt.title("Parabola: (y - p)^2 = 4a(x - q)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.axis("equal")
plt.legend()

plt.show()
```

10. Plot a circle with center at (a, b) and radius r. The values of a=b=2, and r=4, using a while loop.

```python
import numpy as np
import matplotlib.pyplot as plt

a = 2
b = 2
r = 4

theta = 0
step = 0.05

x = []
y = []

while theta <= 2*np.pi:
    x.append(r*np.cos(theta) + a)
    y.append(r*np.sin(theta) + b)
    theta += step

plt.plot(x, y)
plt.scatter(a, b, color='red')
plt.title("Circle using While Loop")
plt.axis("equal")
plt.grid(True)

plt.show()
```

11. Plot the above circle and the ellipse on the same axes. This is give an idea about the intersection points of the two conic sections.

```python
import numpy as np
import matplotlib.pyplot as plt

theta = np.linspace(0, 2*np.pi, 500)

# Circle
xc = 4*np.cos(theta) + 2
yc = 4*np.sin(theta) + 2

# Ellipse
xe = 2 + 4*np.cos(theta)
ye = 2 + 6*np.sin(theta)

plt.plot(xc, yc, label="Circle")
plt.plot(xe, ye, label="Ellipse")

plt.title("Circle and Ellipse")
plt.axis("equal")
plt.grid(True)
plt.legend()

plt.show()
```
