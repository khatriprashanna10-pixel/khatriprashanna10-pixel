1. Write a Python function that takes a list and returns a new list with distinct elements from the
first list.
Sample List : [1,2,3,3,3,3,4,5]
Unique List : [1, 2, 3, 4, 5]


```python
def unique_list(lst):
    unique = []
    for item in lst:
        if item not in unique:
            unique.append(item)
    return unique
numbers = [1,2,3,3,3,3,4,5]
print(unique_list(numbers))
```

2. Write a Python function that takes a number as a parameter and checks whether the number
is prime or not.

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
print(is_prime(13))
```

3. Write a program to print twin primes less than 1000. If two consecutive odd numbers are
both prime then they are known as twin primes.


```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
for i in range(3, 1000, 2):
    if is_prime(i) and is_prime(i + 2):
        print(i, i + 2)
```

4. A positive integer is called a perfect number if it is equal to the sum of all of its divisors,
including 1 but excluding the number itself. For example, 6 = 1 + 2 + 3. Write a
function sumOfDivisors that takes a positive integer n and returns the sum of all its divisors
(excluding n).


```python
def sumOfDivisors(n):
    total = 0
    for i in range(1, n):
        if n % i == 0:
            total = total + i
    return total
print(sumOfDivisors(6))
```

7. Write a function that converts a decimal number to binary number.

```python
def decimal_to_binary(num):
    binary = ""
    while num > 0:
        remainder = num % 2
        binary = str(remainder) + binary
        num = num // 2
    return binary
print(decimal_to_binary(10))
```

8. Write a function prodDigits() that inputs a number and returns the product of digits of that
number.

```python
def prodDigits(num):
    product = 1
    while num > 0:
        digit = num % 10
        product = product * digit
        num = num // 10
    return product
print(prodDigits(234))
```

9. Write a function that will take in an unknown number of arguments and multiply all of them
together and run the function for these 2 sets of numbers:
a. 1,2,3,4,5
b. 12,13,14


```python
def multiply(*args):
    product = 1
    for i in args:
        product = product * i
    return product
print(multiply(1,2,3,4,5))
print(multiply(12,13,14))
```

Set
A Set in Python programming is an unordered, unchangeable*, and unindexed collection data
type that is inerrable, mutable and has no duplicate elements.
*Note: Set items are unchangeable, but you can remove items and add new items.

Example 3:
Perform the following operations on given sets.
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}
set3={10,20,30}


a. Find the length of set1


```python
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}
set3 = {10, 20, 30}

# a. Length of set1
print("Length of set1 =", len(set1))

# b. Sum of elements in set2
print("Sum of set2 =", sum(set2))

# c. Union
print("Union =", set1.union(set2))

# d. Intersection
print("Intersection =", set1.intersection(set2))

# e. set1 - set2
print("set1 - set2 =", set1 - set2)

# f. set2 - set1
print("set2 - set1 =", set2 - set1)

# g. Check subset
print("Is set3 subset of set1?", set3.issubset(set1))
print("Is set3 subset of set2?", set3.issubset(set2))

# h. Complement of set1
U = set(range(1, 51))
print("Complement of set1 =", U - set1)

# i. Check if 15 is present
print("Is 15 present in set1?", 15 in set1)

# j. Add 60 to set1
set1.add(60)
print("After adding 60 to set1 =", set1)

# k. Remove 60 from set2
set2.remove(60)
print("After removing 60 from set2 =", set2)

# l. Remove all elements from set3
set3.clear()
print("set3 after clearing =", set3)

# m. Maximum and minimum value in set2
print("Maximum value in set2 =", max(set2))
print("Minimum value in set2 =", min(set2))
```
