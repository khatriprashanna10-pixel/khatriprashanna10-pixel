a. Create three vectors in 3D space and plot them using Matplotlib.
U= np.array([1, 0.25 ,0])
V = np.array([0, 1, 0.25])
W = np.array([0.25, 0, 1])
To do 3D plot, you will need to create 3D axes using following code:
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d'

```python
import numpy as np
import matplotlib.pyplot as plt

U=np.array([1,0.25,0])
V=np.array([0,1,0.25])
W=np.array([0.25,0,1])

fig=plt.figure()
ax=fig.add_subplot(111,projection='3d')

origin=[0,0,0]

ax.quiver(origin[0],origin[1],origin[2],
          U[0],U[1],U[2],color='r')

ax.quiver(origin[0],origin[1],origin[2],
          V[0],V[1],V[2],color='g')

ax.quiver(origin[0],origin[1],origin[2],
          W[0],W[1],W[2],color='b')

plt.show()
```

b. Create two vectors in 2D space, and add them to create a third vector, and plot all three vectors
using Matplotlib.
U = np.array([-0.5, 0.5])
V = np.array([1,1])
The two vectors are added as
W = U + V # adding two vector

```python
import numpy as np
import matplotlib.pyplot as plt

U=np.array([-0.5,0.5])
V=np.array([1,1])
W=U+V

origin=np.array([0,0])

plt.quiver(origin[0],origin[1],U[0],U[1],angles='xy',scale_units='xy',scale=1,color='r')
plt.quiver(origin[0],origin[1],V[0],V[1],angles='xy',scale_units='xy',scale=1,color='b')
plt.quiver(origin[0],origin[1],W[0],W[1],angles='xy',scale_units='xy',scale=1,color='g')

plt.xlim(-2,3)
plt.ylim(-2,3)
plt.grid()
plt.show()
```

Write a function to calculate magnitude of a vector. For example, calculate the magnitude of
the vector be U = (1.5, 1.5, 2.0)

```python
import numpy as np

def magnitude(v):
    return np.linalg.norm(v)

U=np.array([1.5,1.5,2.0])

print(magnitude(U))
```

d. Write a function to calculate the direction cosines of the vector. For example, let the vector be
O = (1.5, 1.5, 2.0).

```python
import numpy as np

def direction_cosines(v):
    mag=np.linalg.norm(v)
    return v/mag

O=np.array([1.5,1.5,2.0])

print(direction_cosines(O))
```

e. Write a function to calculate Scalar (dot) product of two vectors. For example, let the vectors
be a= (1.5, 1.5, 2.0) and b = (−1.5, 0.5, 2.0)

```python
import numpy as np

def dot_product(a,b):
    return np.dot(a,b)

a=np.array([1.5,1.5,2.0])
b=np.array([-1.5,0.5,2.0])

print(dot_product(a,b))
```

f. Write a function to calculate Vector (cross) product of two vectors. For example, let the vectors
be a = (1.5, 1.5, 2.0) and b= (−1.5, 0.5, 2.0)

```python
import numpy as np

def cross_product(a,b):
    return np.cross(a,b)

a=np.array([1.5,1.5,2.0])
b=np.array([-1.5,0.5,2.0])

print(cross_product(a,b))
```
