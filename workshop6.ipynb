Problem 2: Given a matrix and a vector, write a Python function that performs Matrix-Vector
product. Check your result with NumPy built-in function to perform Matrix-Vector/Matrix-
Matrix product.

```python
import numpy as np

matrix = np.array([[3,3,3],
                   [3,3,3],
                   [3,3,3]])

vector = np.array([[2],
                   [2],
                   [2]])

print("Matrix =")
print(matrix)

print("Vector =")
print(vector)

# Method 1
result1 = np.matmul(matrix, vector)

print("Result1 =")
print(result1)

# Method 2
result2 = np.dot(matrix, vector)

print("Result2 =")
print(result2)
```

Problem 3: Given two compatible matrices, write a Python function that performs Matrix-Matrix
product. Check your result with NumPy built-in function to perform Matrix-Vector/Matrix-
Matrix product.

```python
def matrix_matrix_product(A, B):

    rows = len(A)
    cols = len(B[0])
    common = len(B)

    result = []

    for i in range(rows):
        row = []
        for j in range(cols):
            total = 0
            for k in range(common):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)

    return result

A = [[1,2],
     [3,4]]

B = [[5,6],
     [7,8]]

print("Manual:")
print(matrix_matrix_product(A,B))

A_np = np.array(A)
B_np = np.array(B)

print("\nNumPy:")
print(np.dot(A_np,B_np))
```

Problem 4: Given a matrix, write a Python function that checks if it is a symmetric matrix

```python
def is_symmetric(matrix):
    matrix = np.array(matrix)
    return np.array_equal(matrix, matrix.T)

A = np.array([[1,2,3],
              [2,5,6],
              [3,6,9]])

B = np.array([[1,2],
              [3,4]])

print("A is symmetric:", is_symmetric(A))
print("B is symmetric:", is_symmetric(B))
```
